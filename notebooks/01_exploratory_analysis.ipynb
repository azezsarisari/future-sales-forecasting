# Future Sales Forecasting
## Exploratory Data Analysis

Dataset: Kaggle Store Sales - Time Series Forecasting

Objective: Analyze historical daily sales patterns and support a 30-day forecasting pipeline.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_PATH = PROJECT_ROOT / 'data/processed/daily_sales.csv'
df = pd.read_csv(DATA_PATH, parse_dates=['date'])
df.head()
```

## 1. Dataset Overview

```python
print('Rows:', len(df))
print('Date range:', df['date'].min(), 'to', df['date'].max())
print('Missing sales:', df['sales'].isna().sum())
display(df.describe())
```

## 2. Historical Sales Trend

```python
fig, ax = plt.subplots(figsize=(14, 5))
ax.plot(df['date'], df['sales'], linewidth=0.8)
ax.set_title('Historical Daily Sales')
ax.set_xlabel('Date')
ax.set_ylabel('Sales')
ax.grid(alpha=0.25)
plt.show()
```

## 3. Weekly Seasonality

```python
weekly = df.copy()
weekly['weekday'] = weekly['date'].dt.day_of_week
weekday_avg = weekly.groupby('weekday')['sales'].mean()

fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'], weekday_avg.reindex(range(7)))
ax.set_title('Average Sales by Weekday')
ax.set_ylabel('Average Sales')
plt.show()
```

## 4. Monthly Seasonality

```python
monthly = df.copy()
monthly['month'] = monthly['date'].dt.month
monthly_avg = monthly.groupby('month')['sales'].mean()

fig, ax = plt.subplots(figsize=(10, 5))
ax.bar(monthly_avg.index, monthly_avg.values)
ax.set_title('Average Daily Sales by Month')
ax.set_xlabel('Month')
ax.set_ylabel('Average Daily Sales')
plt.show()
```

## 5. Missing Dates

```python
missing = df[df['is_missing_date'] == 1]
display(missing[['date', 'sales', 'is_missing_date']])
```

## 6. Final Model Performance

```python
metrics_path = PROJECT_ROOT / 'reports/metrics/final_test_metrics.json'
metrics = pd.read_json(metrics_path, typ='series')
display(metrics)
```

- Historical sales show an overall upward trend.
- Weekly seasonality is visible, with higher weekend sales.
- Four calendar dates are missing from the historical records.
